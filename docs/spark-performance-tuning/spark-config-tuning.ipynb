```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .config("spark.driver.memory", "10g")
    .appName("performnce-tuner")
    .getOrCreate()
)
```

#### 1. Optimal Executor Sizing

- Number of Executors `spark.executor.instances`
- Memory per Executors `spark.executor.memory`

```python
MEMORY_OVERHEAD = ""
EXECUTOR_MEMORY = "10M"
```

#### 2. Memory Management

```python
# On-Heap Memory
spark.conf.set("spark.executor.memory", EXECUTOR_MEMORY)

# Unified Memory
spark.conf.set("spark.memory.fraction", "0.6")

# Storage Memory
spark.conf.set("spark.memory.storageFraction", "0.5")

# Off-Heap Memory
spark.conf.set("spark.memory.offHeap.enabled", "true")
spark.conf.set("spark.memory.offHeap.size", str(0.1 * EXECUTOR_MEMORY))

# Overhead Memory
spark.conf.set("spark.executor.memoryOverhead", MEMORY_OVERHEAD)
```

#### 3. Broadcast Joins

```python
# Increase threshold — broadcast tables up to 100MB automatically
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", 100 * 1024 * 1024)  # 100MB

# Disable auto broadcast entirely
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)
```
